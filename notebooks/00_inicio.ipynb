# Plantar la semilla

```python
def my_agent(obs):
    market = []
    if obs["step"] == 0:
        market = [["BUY_SEED", "WHEAT", 1]]

    farmer_pos = obs["farms"][obs["player"]]["farmer"]
    x, y = farmer_pos
    tile = obs["farms"][obs["player"]]["tiles"][y][x]
    wheat_seeds = obs["private"]["seeds"].get("WHEAT", 0)

    if tile is None and wheat_seeds > 0: # Comprueba que no hay nada en la celda y que hay semilla en inventario
        farmer = ["PLANT", "WHEAT"] # Se planta la semilla
    else:
        farmer = ["PASS"]

    return {"farmer": farmer, "hands": [], "market": market}
```

# Regar la semilla

```python
def my_agent(obs):
    market = []
    if obs["step"] == 0:
        market = [["BUY_SEED", "WHEAT", 1]]

    farmer_pos = obs["farms"][obs["player"]]["farmer"]
    x, y = farmer_pos
    tile = obs["farms"][obs["player"]]["tiles"][y][x]
    #print(tile)
    wheat_seeds = obs["private"]["seeds"].get("WHEAT", 0)

    if tile is None and wheat_seeds > 0: # Comprueba que no hay nada en la celda y que hay semilla en inventario
        farmer = ["PLANT", "WHEAT"] # Se planta la semilla
    elif tile is not None and tile.get("kind") == "PLANT" and tile.get("watered_today") == False: # Comprobación de si se ha regado
        farmer = ["WATER"] # Se riega la semilla
    else:
        farmer = ["PASS"]

    return {"farmer": farmer, "hands": [], "market": market}
```

# Cosechar la semilla

```python
def my_agent(obs):
    market = []
    if obs["step"] == 0:
        market = [["BUY_SEED", "WHEAT", 1]]

    farmer_pos = obs["farms"][obs["player"]]["farmer"]
    x, y = farmer_pos
    tile = obs["farms"][obs["player"]]["tiles"][y][x]
    #print(tile)
    wheat_seeds = obs["private"]["seeds"].get("WHEAT", 0)

    if tile is None and wheat_seeds > 0: # Comprueba que no hay nada en la celda y que hay semilla en inventario
        farmer = ["PLANT", "WHEAT"] # Se planta la semilla
    elif tile is not None and tile.get("kind") == "PLANT" and tile.get("watered_today") == False: # Comprobación de si se ha regado
        farmer = ["WATER"] # Se riega la semilla
    elif tile is not None and tile.get("kind") == "PLANT" and obs["day"] - tile["planted_day"] >= 2: # Comprueba si han pasado más de 2 días desde que se plantó para cosechar.
        farmer = ["HARVEST"] # Se cosecha la semilla
    else: 
        farmer = ["PASS"]

    return {"farmer": farmer, "hands": [], "market": market}
```

# Venta del trigo

```python
def my_agent(obs):
    market = []
    if obs["step"] == 0:
        market = [["BUY_SEED", "WHEAT", 1]]

    farmer_pos = obs["farms"][obs["player"]]["farmer"]
    x, y = farmer_pos
    tile = obs["farms"][obs["player"]]["tiles"][y][x]
    #print(tile)
    wheat_seeds = obs["private"]["seeds"].get("WHEAT", 0) # Revisa si existen semillas en el inventario
    wheat_in_shed = obs["private"]["shed"].get("WHEAT", 0) # Revisa si existe trigo en el inventario
    if wheat_in_shed > 0: # Si hay trigo lo vende en el mercado
        market = [["SELL","WHEAT",1]]

    if tile is None and wheat_seeds > 0: # Comprueba que no hay nada en la celda y que hay semilla en inventario
        farmer = ["PLANT", "WHEAT"] # Se planta la semilla
    elif tile is not None and tile.get("kind") == "PLANT" and tile.get("watered_today") == False: # Comprobación de si se ha regado
        farmer = ["WATER"] # Se riega la semilla
    elif tile is not None and tile.get("kind") == "PLANT" and obs["day"] - tile["planted_day"] >= 2: # Comprueba si han pasado más de 2 días desde que se plantó para cosechar.
        farmer = ["HARVEST"] # Se cosecha la semilla
    else: 
        farmer = ["PASS"]
    return {"farmer": farmer, "hands": [], "market": market}
```

Bucle de trigo construido paso a paso contra `"random"`, añadiendo una pieza cada vez y comprobando el efecto antes de seguir:

| Paso | Qué se añadió | Resultado observado |
|---|---|---|
| Base | `PASS` siempre | 3000.0 vs 0.0 |
| Compra semilla | `BUY_SEED` en `step == 0` | 2990.0 vs 0.0 (gasto sin retorno) |
| Plantar | `PLANT` si tile vacío y hay semilla | tile termina en `WEED` (nunca se riega) |
| Regar | `WATER` si `watered_today == False` | tile sigue acabando en `WEED` (nunca se cosecha; decae tras `max_lifespan_step`) |
| Cosechar | `HARVEST` si `obs["day"] - planted_day >= 2` | trigo pasa al `shed`, aún sin vender |
| Vender | `SELL` cuando hay trigo en `shed` | 3045.0 vs 0.0 (un solo ciclo plantar-cosechar-vender) |
| Recomprar semilla | condición `wheat_seeds <= 0` en vez de solo `step == 0` | ~3800-3900 vs 0.0 (el ciclo se repite solo durante toda la partida) |

Bugs encontrados y corregidos por el camino (quedan documentados aquí para no repetirlos):
- `tile is {"kind": "PLANT"}` no sirve para comparar diccionarios (compara identidad de objeto, y además el `tile` real tiene más claves que solo `"kind"`) — la forma correcta es `tile.get("kind") == "PLANT"`.
- Llamar a `tile.get(...)` cuando `tile` puede ser `None` lanza `AttributeError: 'NoneType' object has no attribute 'get'` — hace falta comprobar `tile is not None` antes.
- El orden de las líneas importa: una variable debe estar definida *antes* de usarse (`wheat_seeds` se calculaba después del `if` que ya la necesitaba, moverla arriba lo arregló).
